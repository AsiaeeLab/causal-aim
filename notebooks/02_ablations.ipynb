# 2. Ablations and the workload-dimension figure

The `--all` run in notebook 1 already produced the ablation CSVs. This notebook adds
the one extra run the workload-dimension figure needs (SNR thresholding on), then
draws all five ablation and fidelity figures.


## The thresholded workload-dimension run

The dimension figure compares the pipeline with and without SNR thresholding. The
`--all` run gives the unthresholded arm; this adds the thresholded arm at tau=3.


```python
import subprocess
cmd = ('python run_experiments.py --ablations dim --replications-ablation 200 '
       '--calibration-snr-threshold 3.0 --use-real-data --ihdp-npz-path data/raw/ihdp_npci_1-100.merged.npz --twins-csv-path data/raw/twins.csv --acic-dir data/raw/acic --lalonde-csv-path data/raw/lalonde.csv '
       '--results-dir results_dim_snr3 --figures-dir results_dim_snr3/figs --n-jobs 20')
subprocess.run(cmd.split(), cwd='..', check=True)
```

## Draw the ablation figures
