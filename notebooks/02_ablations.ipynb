# 2. Ablations and the workload-dimension figure

Notebook 1's `--all` run already produced the ablation CSVs. This notebook adds the one
extra run the workload-dimension figure needs and redraws the five ablation and fidelity
figures. Run notebook 1 first.


## Step 1 - the SNR-thresholded workload-dimension run

The workload-dimension figure compares the pipeline with and without SNR thresholding.
The `--all` run gives the unthresholded curve; this run adds the thresholded curve at
threshold 3. It is a single ablation, so it finishes quickly.


```python
import subprocess
cmd = [
    'python', 'run_experiments.py', '--ablations', 'dim',
    '--replications-ablation', '200', '--calibration-snr-threshold', '3.0',
    '--use-real-data', '--ihdp-npz-path', 'data/raw/ihdp_npci_1-100.merged.npz', '--twins-csv-path', 'data/raw/twins.csv', '--acic-dir', 'data/raw/acic', '--lalonde-csv-path', 'data/raw/lalonde.csv',
    '--results-dir', 'results_dim_snr3', '--figures-dir', 'results_dim_snr3/figs',
    '--n-jobs', '20',
]
subprocess.run(cmd, cwd='..', check=True)
```
